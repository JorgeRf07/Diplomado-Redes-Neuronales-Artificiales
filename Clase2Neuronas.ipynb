```python
#!pip install tensorflow
#Cargamos librerias
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense#dense es la neurona
from tensorflow.keras.optimizers import SGD#SGD es el gradiente descendente
```

```python
X = tf.constant([[1.0,2.0,3.0]])
y = tf.constant([[1.0]])
print(type(X))
```

```python
#Modelo 2 Neurona o capas ocultas en medio y 1 capa o neurona de salida
modelo = Sequential([Dense(2, activation='sigmoid',input_shape=(3,)),
                     Dense(1,activation="sigmoid")])
#SGD => Gradiente Descendente =>Backpropagation
modelo.compile(optimizer=SGD(learning_rate = 0.12),loss = "mse")
#Ver pesos iniciales
print("Pesos Iniciales")
for capa in modelo.layers:
  print(capa.get_weights())

#Mostramos una sola actualizacion
#Entrenamos nuestro modelo para que aprenda
modelo.fit(X,y,epochs=1,verbose = 1)#Epochs son la cantidad de iteraciones
#verbose va a detallar la informacion

#ver pesos despues de la actualizacion
print("\nPesos Actualizados\n")
for capa in modelo.layers:
  print(capa.get_weights())

  #Prediccion con respecto a las entradas
  prediccion = modelo.predict(X)
  print("Prediccion")
  print(prediccion)
```

```python
import tensorflow as tf
import numpy as np

x = tf.constant([[1.0,2.0,3.0]])
y = tf.constant([[1.0]])

modelo = Sequential([Dense(2, activation='sigmoid',input_shape=(3,)),
                     Dense(1,activation='sigmoid')])

W = np.array([[0.2,0.3],[0.4,0.5],[0.1,0.2]])
b = np.array([0.1,0.2])
V = np.array([[0.8],[0.7]])
b3 = np.array([0.1])

#Asignar pesos a las capas
modelo.layers[0].set_weights((W,b))
modelo.layers[1].set_weights((V,b3))

#Compilamos el modelo
modelo.compile(optimizer = SGD(learning_rate=0.1), loss = "mse")

print("prediccion antes")
print(modelo.predict(X))
modelo.fit(X,y,epochs=1, verbose = 1)
print("prediccion despues")
print(modelo.predict(X))

for i, capa in enumerate(modelo.layers):
  print("\nnCapa,i+1")
  print(capa.get_weights)
```
