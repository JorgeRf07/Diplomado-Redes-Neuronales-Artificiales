```python
#0.8 y 0.5 para gato >=0.5
#0.3 para perro <0.5
#Forward propagation
#Back propagation -> disminuir el error
entradas = [0.8,0.3,0.5]
pesos=[0.4,0.7,0.2]
sesgo=0.1
#suma(xi*w1)+b
# y_p = (x1*w1+w2*w2+x3*w3)+b
y_p=(entradas[0]*pesos[0]+entradas[1]*pesos[1]+entradas[2]*pesos[2])+sesgo
print(f"y_p = {y_p}")
if(y_p >= 0.5):
  print ("gato")
else:
  print("perro")
```
