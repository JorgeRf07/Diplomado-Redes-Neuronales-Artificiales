```python


import pandas as pd#//////esta libreria se importa(utilizada para manipular y analizar datos)
#y asigna el alias pd
import matplotlib.pyplot as plt#importa el modulo pylot de Matplotlib(para crear graficos) con el
#alas plt
URL = 'https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv'
df = pd.read_csv(URL)#Descarga y lee el archivo CSV desde la URL, guardando la tabla de datos
#dentro de una variable DataFrame o df
df.head()#Devuelve por defecto  las () primeras filas del DataFrame aunq no se muestra
print(df.head(30))#Imprime las (8) filas de la tabla de datos
```

```python
import seaborn as sns
plt.figure(figsize=(12, 7))
plt.title('Distribución de Edades del Titanic por Sexo')
plt.xlabel('Edad')
plt.ylabel('Frecuencia')
plt.xticks(range(0, 90, 10))
plt.grid(True)
sns.histplot(data=df, x='Age', hue='Sex', bins=20, kde=True, palette=
{'male': 'skyblue', 'female': 'lightcoral'})
plt.legend(title='Sexo', labels=['Mujer', 'Hombre'])
plt.show()
```
