```python
import math

#Entradas, pesos y sesgo

x= [1.5, 0.8, -0.4]
w= [0.6, -0.3, 0.9]
b= 0.1

#Suma ponderada

z = sum(xi * wi for xi, wi in zip(x,w)) + b
print(f"Suma ponderada z = {z:.4f}")

# Funcion sigmoide
def sigmoide(z):
  return 1 / (1 + math.exp(-z))

  activacion = sigmoide(z)
  print(f"Activacion õ(z) = {activacion:.4f}")


#Clasificacion por umbral
clase = 1 if activacion >= 0.5 else 0
print(f"Clase predicha: {clase}")
```
